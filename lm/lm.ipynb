```python
# Set a fixed random seed for reproducibility
torch.manual_seed(42)

# Fixed input tensor parameters
batch_size = 2
height = 4
width = 4
channels = 3
pixels = height * width

# Fixed input tensors
channels_last = torch.randn(batch_size, height, width, channels, dtype=torch.float64) # Use float64 for higher precision
B = torch.randn(pixels, pixels, dtype=torch.float64) # Use float64 for higher precision

# --- Torch approach ---
print("--- Running Torch Approach ---")
# 1. (batch, height, width, channel) -> (batch, H*W, channel)
channels_last_flat_torch = channels_last.view(
    batch_size, pixels, channels
)
print(f"channels_last_flat_torch shape: {channels_last_flat_torch.shape}")

# 2. (batch, H*W, channel) -> (batch, channel, H*W)
channels_first_flat_torch = channels_last_flat_torch.transpose(1, 2)
print(f"channels_first_flat_torch shape: {channels_first_flat_torch.shape}")

# 3. Apply transformation: (batch, channel, H*W_in) @ (H*W_in, H*W_out) -> (batch, channel, H*W_out)
# Note: B has shape (pixel_out, pixel_in). For matmul, we need (pixel_in, pixel_out)
# So, channels_first_flat_torch @ B.T is correct.
channels_first_transformed_torch = channels_first_flat_torch @ B.T
print(f"channels_first_transformed_torch shape: {channels_first_transformed_torch.shape}")

# 4. (batch, channel, H*W) -> (batch, H*W, channel)
channels_last_flat_transformed_torch = channels_first_transformed_torch.transpose(1, 2)
print(f"channels_last_flat_transformed_torch shape: {channels_last_flat_transformed_torch.shape}")

# 5. (batch, H*W, channel) -> (batch, height, width, channel)
channels_last_transformed_torch = channels_last_flat_transformed_torch.view(*channels_last.shape)
print(f"Final Torch output shape: {channels_last_transformed_torch.shape}")


# --- Einops approach ---
print("\n--- Running Einops Approach ---")
# 1. (batch, height, width, channel) -> (batch, channel, H*W)
channels_first_einops = rearrange(
    channels_last,
    "batch height width channel -> batch channel (height width)"
)
print(f"channels_first_einops shape: {channels_first_einops.shape}")

# 2. Apply transformation using einsum
# "batch channel pixel_in, pixel_out pixel_in -> batch channel pixel_out"
# This means: take 'pixel_in' from the first tensor, 'pixel_in' and 'pixel_out' from B.
# Contract 'pixel_in'. Result contains 'batch', 'channel', 'pixel_out'.
channels_first_transformed_einops = einsum(
    channels_first_einops, B,
    "batch channel pixel_in, pixel_out pixel_in -> batch channel pixel_out"
)
print(f"channels_first_transformed_einops shape: {channels_first_transformed_einops.shape}")

# 3. (batch, channel, H*W) -> (batch, height, width, channel)
channels_last_transformed_einops = rearrange(
    channels_first_transformed_einops,
    "batch channel (height width) -> batch height width channel",
    height=height, width=width
)
print(f"Final Einops output shape: {channels_last_transformed_einops.shape}")

# --- Comparison ---
# Using a higher tolerance (e.g., 1e-8 or 1e-9) might still be true
# since we used float64, but 1e-6 is generally safe for float32 as well.
are_same = torch.allclose(channels_last_transformed_torch, channels_last_transformed_einops, atol=1e-8)

print(f"\nAre the final results numerically identical (within tolerance of 1e-8)? {are_same}")

# Optional: Print a small portion to visually inspect differences if any
# if not are_same:
#     print("\nDifferences found:")
#     diff = torch.abs(channels_last_transformed_torch - channels_last_transformed_einops)
#     print(diff.max())
#     print(torch.where(diff > 1e-8)) # Show indices where difference is significant
```
